```r
ruta_exportar <- "/content/drive/MyDrive/4Students_2026/Data/Selva/Downscalling/"  #pongan el / al final
```

## Datos historicos

```r
df <- read.table(
  file = archivos[estacion_numero],
  sep = ",",
  header = TRUE
)
df$Fecha <- as.Date(df$Fecha)
df
```

```r
date_cal<-seq(as.Date("1981-01-01"),as.Date("2000-12-31"),by="day")
date_val<-seq(as.Date("2001-01-01"),as.Date("2014-12-31"),by="day")
```

```r
o1 <-list();p1 <-list();s1 <-list();
o1[[1]]<-df$Pp_obs[df$Fecha %in% date_cal] #DATA OBSERVADA - CALIBRACION
p1[[1]]<-df$PR_ssp245[df$Fecha %in% date_cal] #DATA HISTORICA DEL MODELO - CALIBRACION (1981-2000)
s1[[1]]<-df$PR_ssp245[df$Fecha %in% date_val] #validacion (2000-2014)
```

```r
plot(date_val,df$Pp_obs[df$Fecha %in% date_val],col="black",type="l",lty=1,xlab="Fechas",ylab="Pp")
```

```r
data_bc1<-list()
data_bc1[[1]]<-as.vector(downscaleR:::biasCorrection1D(o=o1,p=p1,s=s1,
                                                      method="eqm",
                                                      pr.threshold=0.1,
                                                      n.quantiles = NULL,
                                                      extrapolation = "constant",
                                                      precip=TRUE))

plot(date_val,s1[[1]],type="l",lty=1,col="blue",xlab="Fechas",ylab="Pp") #datos

lines(date_val,df$Pp_obs[df$Fecha %in% date_val],col="black") #Data historica - validacion
lines(date_val,data_bc1[[1]],col="red")
legend(date_val[1], 20, legend=c("Raw", "Obs","Corrected"),col=c("blue","black", "red"), lty=1:2, cex=0.8)
```

```r
data_bc1[[2]]<-as.vector(downscaleR:::biasCorrection1D(o=o1,p=p1,s=s1,
                                                        method="scaling",
                                                        scaling.type="multiplicative",
                                                        extrapolation = "constant",
                                                        precip=TRUE))


#QUANTILE DATA MAPPING
data_bc1[[3]]<-as.vector(downscaleR:::biasCorrection1D(o=o1,p=p1,s=s1,
                                                        method="qdm",
                                                        pr.threshold=0.1,
                                                        n.quantiles = NULL,
                                                        extrapolation = "constant",
                                                        precip=TRUE))
```

```r
obs1 <-df$Pp_obs[df$Fecha %in% date_val]
df_taylor_val1<-data.frame(date=date_val,obs=obs1, eqm=data_bc1[[1]],
                          scaling = data_bc1[[2]], qdm= data_bc1[[3]])
df_taylor_val1
```

```r
write.table(df_taylor_val1,paste0(ruta_exportar,"Downscalling_hist_pp_",nombre_corto), sep = ",", row.names = FALSE) #guardo las metricas
```

## Datos futuros ssp 2 - 4.5

```r
date_calF<-seq(as.Date("1981-01-01"),as.Date("2014-12-31"),by="day")
date_ssp <-seq(as.Date("1981-01-01"),as.Date("2100-12-31"),by="day")
```

```r
o2<-list();p2<-list();s2<-list();
o2[[1]]<-df$Pp_obs[df$Fecha %in% date_calF] #Data historica 1981-2014
p2[[1]]<-df$PR_ssp245[df$Fecha %in% date_calF] #Data historica del modelo 1981-2014
s2[[1]]<-df$PR_ssp245[df$Fecha %in% date_ssp]  #Data futura 2015-2100

data_bc2 <-list()
data_bc2[[1]]<-as.vector(downscaleR:::biasCorrection1D(o=o2,p=p2,s=s2,
                                                        method="eqm",
                                                        pr.threshold=0.1,
                                                        n.quantiles = NULL,
                                                        extrapolation = "constant",
                                                        precip=TRUE))

data_bc2[[2]]<-as.vector(downscaleR:::biasCorrection1D(o=o2,p=p2,s=s2,
                                                        method="scaling",
                                                        scaling.type="multiplicative",
                                                        extrapolation = "constant",
                                                        precip=TRUE))


#QUANTILE DATA MAPPING
data_bc2[[3]]<-as.vector(downscaleR:::biasCorrection1D(o=o2,p=p2,s=s2,
                                                        method="qdm",
                                                        pr.threshold=0.1,
                                                        n.quantiles = NULL,
                                                        extrapolation = "constant",
                                                        precip=TRUE))
```

```r
length(date_ssp)
length(df$PR_ssp245[df$Fecha %in% date_ssp])
```

```r
length(data_bc2[[3]])
```

```r
plot(df$Fecha[df$Fecha %in% date_ssp],df$PR_ssp245[df$Fecha %in% date_ssp] ,col="black")
```

```r
# 1. Creamos el primer gráfico (el lienzo)
plot(df$Fecha[df$Fecha %in% date_ssp],df$PR_ssp245[df$Fecha %in% date_ssp], type = "l",
      lty = 2, col = "black")

# 2. Agregamos la segunda serie sobre el mismo gráfico (sin borrar)
lines(df$Fecha[df$Fecha %in% date_ssp], data_bc2[[3]],
     type = "l", lty = 1, col = "red",
     ylim = c(0, 100),
     main = "Comparación de Precipitación",
     xlab = "Fecha", ylab = "PR (mm/día)")

# 3. Agregamos la leyenda
legend("topleft", # Puedes usar coordenadas o palabras clave como "topleft"
       legend = c("Modelo downscalling", "Modelo original"),
       col = c("red", "black"),
       lty = 1:2,
       cex = 0.8,
       bty = "n") # bty="n" quita el recuadro de la leyenda si prefieres
```

```r
df_taylor_futuro1<-data.frame(date=df$Fecha[df$Fecha %in% date_ssp],eqm=data_bc2[[1]],
                          scaling = data_bc2[[2]], qdm= data_bc2[[3]])
df_taylor_futuro1
```

```r
write.table(df_taylor_futuro1,paste0(ruta_exportar,"Downscalling_ssp2_4.5_pp_",nombre_corto), sep = ",", row.names = FALSE) #guardo las metricas
```
